```python
# ==============================================================================
# CÉLULA 1: LEITURA DIRETA DO ARQUIVO JSON UTILIZANDO O PANDAS
# ==============================================================================

# Importa a biblioteca pandas e define o alias padrão mundial 'pd'
import pandas as pd

# Configura o pandas para exibir TODAS as colunas ao imprimir um DataFrame, sem ocultar com '...'
pd.set_option('display.max_columns', None)

# Define o caminho completo do arquivo JSON no disco rígido do computador
# O prefixo 'r' define uma raw string, evitando problemas com caracteres de escape como barras invertidas
caminho_do_arquivo = r"/Users/analeia/Desktop/Engenharia de Dados/V_OCORRENCIA_AMPLA.json"

# pd.read_json lê a estrutura do JSON e a carrega na memória como uma tabela chamada DataFrame (variável 'db')
# O parâmetro encoding='utf-8-sig' trata arquivos codificados em UTF-8 com ou sem marcador de ordem de byte (BOM)
db = pd.read_json(caminho_do_arquivo, encoding='utf-8-sig')

# db.head() exibe por padrão as 5 primeiras linhas da tabela para inspecionar os dados carregados
db.head()
```

```python
# ==============================================================================
# CÉLULA 2: DETECÇÃO AUTOMÁTICA DE ENCODING COM A BIBLIOTECA CHARDET
# ==============================================================================

# Comando mágico do Jupyter (%pip) para instalar a biblioteca chardet no ambiente virtual ativo
%pip install chardet

# Importa o módulo codecs (biblioteca nativa do Python para manipulação de codificações de texto)
import codecs

# Importa a classe UniversalDetector da biblioteca chardet (analisa padrões de bytes para deduzir o charset)
from chardet.universaldetector import UniversalDetector

# Define o caminho do arquivo JSON a ser analisado
caminho_do_arquivo = r"/Users/analeia/Desktop/Engenharia de Dados/V_OCORRENCIA_AMPLA.json"

# Cria uma nova instância do detector universal de codificação
detector = UniversalDetector()

# Abre o arquivo no modo 'rb' (read binary / leitura binária), lendo os bytes puros sem tentar decodificar texto
with open(caminho_do_arquivo, 'rb') as arquivo:
    # Itera linha por linha (em blocos de bytes) sobre o arquivo
    for linha in arquivo:
        # Envia cada linha de bytes para o detector analisar as frequências estatísticas dos caracteres
        detector.feed(linha)
        # Se o detector já atingiu certeza suficiente sobre a codificação, encerra o loop mais cedo para poupar tempo
        if detector.done:
            break

# Encerra o processo de detecção e consolida o resultado
detector.close()

# Extrai a codificação identificada (exemplo: 'utf-8', 'ISO-8859-1')
encoding = detector.result['encoding']

# Exibe o dicionário com o resultado completo: {'encoding': ..., 'confidence': ..., 'language': ...}
print(detector.result)

# Carrega o JSON com o pandas passando dinamicamente o encoding detectado de forma automática
db = pd.read_json(caminho_do_arquivo, encoding=encoding)
```

```python
# ==============================================================================
# CÉLULA 3: FLUXO COMPLETO E CONSOLIDADO DE DETECÇÃO AUTOMÁTICA E LEITURA
# ==============================================================================

# Importa a biblioteca pandas para manipulação de tabelas (DataFrames)
import pandas as pd

# Importa codecs para manipulação de codificações de texto
import codecs

# Importa o UniversalDetector para identificar automaticamente a codificação de arquivos de texto
from chardet.universaldetector import UniversalDetector

# Define o caminho do arquivo JSON com as ocorrências da ANAC
caminho_do_arquivo = r"/Users/analeia/Desktop/Engenharia de Dados/V_OCORRENCIA_AMPLA.json"

# Inicializa o detector universal de codificação
detector = UniversalDetector()

# Abre o arquivo no modo de leitura binária ('rb')
with open(caminho_do_arquivo, 'rb') as arquivo:
    for linha in arquivo:
        detector.feed(linha)  # Alimenta o detector com cada linha de bytes
        if detector.done:      # Se já houver certeza suficiente sobre o encoding, para o loop
            break

# Finaliza a análise do detector
detector.close()

# Obtém o nome da codificação detectada (ex: 'utf-8')
encoding = detector.result['encoding']

# Carrega o arquivo JSON no DataFrame principal 'db' utilizando o encoding detectado
db = pd.read_json(caminho_do_arquivo, encoding=encoding)

# Caminho do arquivo para carregar na variável de demonstração db_jsonauto
caminho_do_arquivo = r"/Users/analeia/Desktop/Engenharia de Dados/V_OCORRENCIA_AMPLA.json"
db_jsonauto = pd.read_json(caminho_do_arquivo, encoding=encoding)

# Exibe as primeiras 5 linhas do DataFrame resultante para conferência dos dados
db_jsonauto.head()
```
