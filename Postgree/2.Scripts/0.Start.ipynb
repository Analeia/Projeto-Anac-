## 1️⃣ O que é e por que criar um Ambiente Virtual (`venv`)?
Um ambiente virtual é uma pasta isolada onde ficam instaladas as bibliotecas específicas deste projeto.
- **Por que usar?** Evita conflitos de versões entre diferentes projetos e protege o Python do seu sistema operacional.

### Passo 1: Criando o ambiente virtual
```bash
# Executa o módulo 'venv' do Python 3 para criar uma pasta chamada '.venv' com um interpretador isolado
python3 -m venv .venv
```

### Passo 2: Ativando o ambiente virtual
- **No macOS / Linux:**
  ```bash
  # Carrega as variáveis de ambiente do venv para o terminal atual
  source .venv/bin/activate
  ```
- **No Windows (PowerShell):**
  ```powershell
  .venv\Scripts\Activate.ps1
  ```
- **No Windows (Prompt de Comando / CMD):**
  ```cmd
  .venv\Scripts\activate.bat
  ```

> **Como saber se está ativo?**
> O prefixo `(.venv)` aparecerá no início da linha do seu terminal.
> Para confirmar qual Python está ativo no macOS: execute `which python` (o retorno deve apontar para `.venv/bin/python`).

## 2️⃣ Atualizar o Gerenciador de Pacotes (`pip`)
O `pip` é o gerenciador padrão para baixar e instalar pacotes no Python. Mantê-lo atualizado evita erros de download:
```bash
# Atualiza o pip para a versão mais recente dentro do ambiente virtual
pip install --upgrade pip
```
