# GUIA DAS BIBLIOTECAS DO PROJETO

Antes de rodar qualquer comando de instalação, certifique-se de que o seu ambiente virtual está ativado:
- **No macOS / Linux:** `source .venv/bin/activate`
- **No Windows:** `.venv\Scripts\activate`

---

## 📦 Bibliotecas utilizadas e o que cada uma faz:

### 1. `psycopg2-binary` (ou `psycopg2`)
```bash
pip install psycopg2-binary
```
- **O que faz:** É o conector oficial entre o Python e o banco de dados PostgreSQL.
- **Por que precisamos:** Permite abrir uma conexão, enviar instruções SQL (como CREATE TABLE, INSERT, SELECT) e receber os dados de volta no Python.
- *Nota:* A versão `psycopg2-binary` é recomendada para desenvolvimento local porque já vem pré-compilada, evitando erros de compilação C no Mac ou Windows.

### 2. `pandas`
```bash
pip install pandas
```
- **O que faz:** É a principal biblioteca de manipulação, limpeza e análise de dados em Python.
- **Por que precisamos:** Ela lê arquivos JSON, CSV ou Excel e os organiza em formato de tabela bidimensional chamada **DataFrame** (com linhas e colunas), permitindo filtrar colunas, renomear campos e tratar dados com facilidade.

### 3. `chardet`
```bash
pip install chardet
```
- **O que faz:** Detecta automaticamente a codificação de caracteres (encoding) de um arquivo de texto/JSON.
- **Por que precisamos:** Evita erros comuns de leitura causados por acentuação ou caracteres especiais da língua portuguesa (ex.: UTF-8, ISO-8859-1, Windows-1252).

### 4. `openpyxl`
```bash
pip install openpyxl
```
- **O que faz:** Motor de suporte que permite ao pandas ler e salvar planilhas do Microsoft Excel modernas (`.xlsx`).

### 5. `ipykernel`
```bash
pip install ipykernel
```
- **O que faz:** Permite que o VS Code conecte o ambiente virtual diretamente às células dos arquivos Jupyter Notebook (`.ipynb`).

---

## 🔄 Como atualizar uma biblioteca já instalada:
```bash
# O parâmetro --upgrade busca a versão mais recente do pacote e faz a atualização
pip install --upgrade pandas
```
