```python
# ==============================================================================
# PADRÃO DE CONEXÃO COM O BANCO DE DADOS POSTGRESQL USANDO PSYCOPG2
# ==============================================================================

# Importa o módulo psycopg2, que é o conector oficial entre Python e PostgreSQL
import psycopg2

# ------------------------------------------------------------------------------
# PASSO 1: DEFINIÇÃO DOS PARÂMETROS DE CONEXÃO
# ------------------------------------------------------------------------------
# dbname: Nome do banco de dados ao qual queremos nos conectar no PostgreSQL
dbname = 'postgres'

# user: Nome do usuário cadastrado no PostgreSQL com permissões de acesso
user = 'analeia'

# password: Senha de autenticação do usuário informado acima
password = '12345'

# host: Endereço IP ou nome do servidor do banco ('localhost' = máquina local)
host = 'localhost'

# port: Porta TCP onde o serviço do PostgreSQL está atendendo (5432 é a porta padrão)
port = '5432'

# ------------------------------------------------------------------------------
# PASSO 2: ABERTURA DA CONEXÃO COM O BANCO DE DADOS
# ------------------------------------------------------------------------------
# A função psycopg2.connect() recebe as credenciais e tenta abrir o canal de comunicação.
# Se tudo estiver correto, ela retorna um objeto de conexão que guardamos na variável 'conn'.
conn = psycopg2.connect(
    dbname=dbname,      # Informa o banco de dados de destino
    user=user,          # Informa o nome do usuário
    password=password,  # Informa a senha
    host=host,          # Informa o endereço do servidor
    port=port           # Informa a porta de comunicação
)

# ------------------------------------------------------------------------------
# PASSO 3: CRIAÇÃO DO CURSOR
# ------------------------------------------------------------------------------
# O cursor é o objeto responsável por enviar instruções SQL e buscar os resultados.
# Pense no cursor como o 'mensageiro' que leva o comando SQL até o banco e traz os dados de volta.
cur = conn.cursor()

# ------------------------------------------------------------------------------
# PASSO 4: EFETIVAÇÃO DAS OPERAÇÕES (COMMIT)
# ------------------------------------------------------------------------------
# O PostgreSQL trabalha com transações seguras. Operações que modificam dados
# (INSERT, UPDATE, DELETE) só são salvas permanentemente no disco quando chamamos commit().
# Se não chamarmos commit(), ao fechar a conexão o banco cancela todas as alterações (rollback).
conn.commit()

# ------------------------------------------------------------------------------
# PASSO 5: FECHAMENTO DOS RECURSOS (BOAS PRÁTICAS)
# ------------------------------------------------------------------------------
# Fecha o cursor para liberar a memória alocada no banco de dados
cur.close()

# Fecha a conexão com o banco de dados para não deixar conexões presas ou abertas no servidor
conn.close()
```
